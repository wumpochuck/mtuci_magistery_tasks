## Задание 1.
Создать однострочную функцию ```F(A, a)```, принимающую в качестве аргументов дискретное
нечёткое множество ```A``` и числовой параметр ```0 < a < 1```, и возвращающую ```a```-срез ```A_a``` множества ```A```. Продемонстрировать.

````python
def F(A, a):
    return [x for x, mu in A.items() if mu >= a]

A = {'item1': 0.15, 'item2': 0.45, 'item3': 0.55, 'item4': 0.75, 'item5': 0.90}
alpha = 0.5

print(f"Заданное нечёткое множество A: {A}")
print(f"F(A, {alpha}) = {F(A, alpha)}")
````

## Задание 2.
Создать функцию ```F(A)```, принимающую в качестве аргумента дискретное нечёткое множество ```A``` и возвращающую ```ближайшее чёткое``` (обычное) множество и евклидово расстояние от этого ближайшего чёткого множества до ```A```.

````python
def F(A):
    return ({x: (1 if mu >= 0.5 else 0) for x, mu in A.items()}, sum((mu - (1 if mu >= 0.5 else 0))**2 for mu in A.values()) ** 0.5)

A = {'item1': 0.15, 'item2': 0.45, 'item3': 0.55, 'item4': 0.75, 'item5': 0.90}

nearest_set, distance = F(A)
print(f"Исходное нечёткое множество A: {A}")
print(f"Ближайшее чёткое множество A_0: {nearest_set}")
print(f"Евклидово расстояние d_e(A): {distance:.4f}")
````
